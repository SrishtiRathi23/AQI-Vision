# 01 — Exploratory Data Analysis

Load the CPCB data (real if present, else synthetic) and explore AQI structure across cities, seasons and hours.

```python
import sys; sys.path.append('..')
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
import config
from src import ingestion

df = ingestion.load_or_generate()
ingestion.summarize_dataset(df)
df.head()
```

```python
# Mean AQI by city — who is dirtiest?
(df.groupby(config.CITY_COL)[config.TARGET_COL].mean()
   .sort_values(ascending=False).plot.barh(figsize=(8,8)))
plt.xlabel('Mean AQI'); plt.title('Mean AQI by city'); plt.tight_layout()
```

```python
# Seasonal cycle: monthly mean AQI for a few cities.
df['month'] = df[config.DATE_COL].dt.month
for city in ['Delhi', 'Chennai', 'Bengaluru']:
    sub = df[df[config.CITY_COL]==city].groupby('month')[config.TARGET_COL].mean()
    plt.plot(sub.index, sub.values, marker='o', label=city)
plt.legend(); plt.xlabel('Month'); plt.ylabel('Mean AQI'); plt.title('Seasonal cycle')
```

```python
# Pollutant correlation heatmap.
present = [c for c in config.POLLUTANT_COLS + [config.TARGET_COL] if c in df.columns]
sns.heatmap(df[present].corr(), annot=True, fmt='.2f', cmap='RdBu_r')
plt.title('Pollutant correlations')
```
