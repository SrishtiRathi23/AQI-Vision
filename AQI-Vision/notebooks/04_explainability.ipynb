# 04 — Explainability (SHAP)

Load the trained best model and explain its predictions globally and for a single row.

```python
import sys; sys.path.append('..')
import json, pandas as pd
import config
from src import models, explain

model = models.load_model(config.BEST_MODEL_FILE)
cols = json.loads(config.FEATURE_LIST_FILE.read_text())
feat = pd.read_parquet(config.PROCESSED_FILE)
X = feat[cols].sample(1000, random_state=42)
mtype = 'linear' if type(model).__name__ == 'Ridge' else 'tree'
sv = explain.compute_shap_values(model, X, mtype)
```

```python
explain.get_top_features(sv, n=10)
```

```python
explain.plot_shap_summary(sv)
```

```python
# Why this particular prediction?
explain.plot_shap_waterfall(sv, 0)
```
